# Resolução dos Experimentos - Aula 02

Este notebook contém a resolução dos 4 experimentos propostos na última página da aula sobre Formação da Imagem.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
import shutil
from PIL import Image # Usaremos para salvar imagens facilmente com matplotlib ou diretamente

%matplotlib inline
```

### Experimento 1: Imagem em Escala de Cinza
Ler um arquivo de imagem em escala de cinza, exibir, somar 100 unidades a cada pixel, exibir o resultado, comparar e gravar o novo arquivo.

```python
# 1. Ler a imagem e converter para escala de cinza
img_color = mpimg.imread('image.png')

# Se a imagem for RGBA, removemos o alpha ou tratamos apenas os 3 primeiros canais
if img_color.shape[2] == 4:
    img_color = img_color[:,:,:3]

# Convertendo para escala de cinza (média dos canais)
img_gray = np.mean(img_color, axis=2)

# Matplotlib imread pode retornar valores entre 0 e 1 se for PNG.
# Vamos garantir que estamos trabalhando com 0-255 para o experimento fazer sentido com "somar 100"
if img_gray.max() <= 1.0:
    img_gray = (img_gray * 255).astype(np.float32)
else:
    img_gray = img_gray.astype(np.float32)

# 2. Somar 100 unidades (com clip para não passar de 255)
img_gray_plus_100 = np.clip(img_gray + 100, 0, 255).astype(np.uint8)
img_gray = img_gray.astype(np.uint8)

# 3. Exibir e comparar
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title('Original (Escala de Cinza)')
axes[1].imshow(img_gray_plus_100, cmap='gray')
axes[1].set_title('Cinza + 100 Unidades')
plt.show()

# 4. Gravar o novo arquivo
plt.imsave('image_exp1_gray_plus_100.png', img_gray_plus_100, cmap='gray')
```

### Experimento 2: Imagem Colorida
Repetir o mesmo processo com uma imagem colorida.

```python
# 1. Ler a imagem colorida
img_color = mpimg.imread('image.png')
if img_color.shape[2] == 4:
    img_color = img_color[:,:,:3]

if img_color.max() <= 1.0:
    img_color = (img_color * 255).astype(np.float32)
else:
    img_color = img_color.astype(np.float32)

# 2. Somar 100 unidades a cada pixel (em todos os canais)
img_color_plus_100 = np.clip(img_color + 100, 0, 255).astype(np.uint8)
img_color = img_color.astype(np.uint8)

# 3. Exibir e comparar
fig, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].imshow(img_color)
axes[0].set_title('Original (Colorida)')
axes[1].imshow(img_color_plus_100)
axes[1].set_title('Colorida + 100 Unidades')
plt.show()

# 4. Gravar o novo arquivo
plt.imsave('image_exp2_color_plus_100.png', img_color_plus_100)
```

### Experimento 3: Processamento em Lote
Repetir o mesmo processo para um conjunto de imagens gravadas em um diretório.

```python
# Setup: Criar diretórios de teste
test_dir = 'imagens_teste'
output_dir = 'imagens_processadas'
os.makedirs(test_dir, exist_ok=True)
os.makedirs(output_dir, exist_ok=True)

# Copiando a imagem original algumas vezes para simular um lote
for i in range(3):
    shutil.copy('image.png', os.path.join(test_dir, f'img_{i}.png'))

print(f"Processando imagens em '{test_dir}'...")

# Processamento em lote
for filename in os.listdir(test_dir):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        filepath = os.path.join(test_dir, filename)
        
        # Ler
        img = mpimg.imread(filepath)
        if len(img.shape) == 3 and img.shape[2] == 4:
            img = img[:,:,:3]
            
        if img.max() <= 1.0:
            img = (img * 255).astype(np.float32)
        else:
            img = img.astype(np.float32)
            
        # Processar
        img_proc = np.clip(img + 100, 0, 255).astype(np.uint8)
        
        # Salvar
        plt.imsave(os.path.join(output_dir, f'proc_{filename}'), img_proc)
        print(f" - {filename} processada e salva.")

print("Processamento em lote concluído.")
```

### Experimento 4: CSV para Imagem
Leia este arquivo csv e converta em uma imagem em escala de cinza. Grave a imagem em formato PNG.

```python
# 1. Ler o arquivo CSV
# Como vimos na prévia, os dados são separados por vírgula
matriz_imagem = np.loadtxt('imagem.csv', delimiter=',')

print(f"Formato da matriz lida: {matriz_imagem.shape}")

# 2. Exibir como imagem em escala de cinza
plt.figure(figsize=(8, 8))
plt.imshow(matriz_imagem, cmap='gray')
plt.title('Imagem gerada a partir do CSV')
plt.colorbar(label='Intensidade')
plt.show()

# 3. Gravar em formato PNG
# Normalizamos para uint8 antes de salvar para garantir compatibilidade
img_csv_uint8 = np.clip(matriz_imagem, 0, 255).astype(np.uint8)
plt.imsave('imagem_do_csv.png', img_csv_uint8, cmap='gray')

print("Arquivo 'imagem_do_csv.png' gerado com sucesso.")
```
